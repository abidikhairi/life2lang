# Protein BPE Tokenizer — PoC
**Goal**: Download 10,000 SwissProt sequences, train a BPE tokenizer (vocab_size=4096), and save it for later merging with the OPT tokenizer.

Pipeline:
1. Download sequences from UniProt (SwissProt)
2. Extract sequences → `sequences.txt`
3. Train BPE tokenizer
4. Inspect & save

## 1. Install dependencies

## 2. Download 10k SwissProt sequences (FASTA)

```python
import requests

# UniProt REST API — SwissProt (reviewed=true), FASTA format, 10k sequences
url = (
    "https://rest.uniprot.org/uniprotkb/stream"
    "?query=reviewed:true"
    "&format=fasta"
    "&size=10000"
)

print("Downloading sequences...")
response = requests.get(url, stream=True)
response.raise_for_status()

with open("swissprot_10k.fasta", "w") as f:
    for chunk in response.iter_content(chunk_size=8192, decode_unicode=True):
        f.write(chunk)

print("Done. Saved to swissprot_10k.fasta")
```

## 3. Extract sequences only → `sequences.txt`

```python
from Bio import SeqIO

sequences = []
with open("swissprot_10k.fasta") as f:
    for record in SeqIO.parse(f, "fasta"):
        sequences.append(str(record.seq))

print(f"Parsed {len(sequences)} sequences")
print(f"Example: {sequences[0][:80]}...")

# Save — one sequence per line (BPE trainer expects line-delimited corpus)
with open("sequences.txt", "w") as f:
    for seq in sequences:
        f.write(seq + "\n")

print("Saved sequences.txt")
```

## 4. Train BPE tokenizer

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Split
from tokenizers import pre_tokenizers

# Build tokenizer
tokenizer = Tokenizer(BPE(unk_token=None))

# Pre-tokenizer: character-level split
# Protein sequences have no whitespace — we split on every character
# so BPE merges start from individual amino acids (A, C, D, E, ...)
tokenizer.pre_tokenizer = pre_tokenizers.Split(
    pattern="",          # empty string = split every char
    behavior="isolated"
)

trainer = BpeTrainer(
    vocab_size=4096,
    min_frequency=2,
    show_progress=True,
    special_tokens=[]   # no special tokens — handled at OPT tokenizer level
)

print("Training BPE tokenizer...")
tokenizer.train(files=["sequences.txt"], trainer=trainer)
print(f"Vocab size: {tokenizer.get_vocab_size()}")
```

## 5. Sanity check

```python
test_seq = sequences[0][:50]
encoding = tokenizer.encode(test_seq)

print(f"Input  : {test_seq}")
print(f"Tokens : {encoding.tokens}")
print(f"IDs    : {encoding.ids}")
print(f"N tokens: {len(encoding.tokens)} for {len(test_seq)} chars")
```

## 6. Inspect vocabulary

```python
vocab = tokenizer.get_vocab()

# Sort by ID
sorted_vocab = sorted(vocab.items(), key=lambda x: x[1])

print("=== First 50 tokens (single amino acids + early merges) ===")
for token, idx in sorted_vocab[:50]:
    print(f"  {idx:5d} | {token}")

print("\n=== Last 10 tokens (longest merges) ===")
for token, idx in sorted_vocab[-10:]:
    print(f"  {idx:5d} | {token}")
```

## 7. Check collisions with OPT vocabulary (preview)

```python
# NOTE: full merge happens in the next notebook
# Here we just flag potential collisions for awareness

from transformers import AutoTokenizer

opt_tokenizer = AutoTokenizer.from_pretrained("facebook/opt-1.3b")
opt_vocab = set(opt_tokenizer.get_vocab().keys())
protein_vocab = set(vocab.keys())

collisions = opt_vocab & protein_vocab
print(f"OPT vocab size    : {len(opt_vocab)}")
print(f"Protein vocab size: {len(protein_vocab)}")
print(f"Collisions        : {len(collisions)}")
print(f"\nSample collisions : {list(collisions)[:30]}")
```

## 8. Save protein tokenizer

```python
import os
os.makedirs("protein_tokenizer", exist_ok=True)
tokenizer.save("protein_tokenizer/tokenizer.json")
print("Saved to protein_tokenizer/tokenizer.json")
print("Next step: merge with OPT tokenizer (Option 1 — sentinel <PROT>/<\\PROT>)")
```
